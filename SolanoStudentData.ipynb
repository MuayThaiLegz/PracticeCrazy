```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sweetviz as sv
import pygwalker as pyg
import plotly.graph_objects as go
import plotly.express as px

pd.set_option('display.max_columns', None)

ethnicity_mapping = {
    0: 'Not reported',
    1: 'American Indian or Alaska Native',
    2: 'Asian',
    3: 'Pacific Islander',
    4: 'Filipino',
    5: 'Hispanic or Latino',
    6: 'African American',
    7: 'White',
    9: 'Two or More Races'
}

gender_mapping = {
    0: 'Not reported',
    1: 'M',
    2: 'F'
}




# Create new columns for latitude and longitude based on the school name
# ['School_Code', 'Year', 'DISTRICT', 'SCHOOL', 'Ethnicity_Code', 'Gender','Latitude','Longitude']
coords = {
    'Vallejo High': (38.1041, -122.2566),
    'Jesse M. Bethel High': (38.1205, -122.2044),
    'MIT Academy': (38.1489, -122.2454),
    'John Finney High (Continuation)': (38.1133, -122.2443),

}

def create_report(df):
  ret =  sv.analyze(df)
  return ret.show_notebook()

# Solano_County_graduates_by_ethnicity_df = pd.read_csv("/content/drive/MyDrive/ClubStrideFolder/StudentData/Solano_County_graduates_by_ethnicity_df.csv")
```

```python
# First dataset
Solano_County_graduates = pd.read_csv("/content/drive/MyDrive/ClubStrideFolder/StudentData/Solano_County_graduates.csv")
Vallejo_graduates = Solano_County_graduates[Solano_County_graduates['DISTRICT']=='Vallejo City Unified']
Vallejo_graduates['Latitude'] = Vallejo_graduates['SCHOOL'].map(lambda school: coords.get(school, (None, None))[0])
Vallejo_graduates['Longitude'] = Vallejo_graduates['SCHOOL'].map(lambda school: coords.get(school, (None, None))[1])
Vallejo_graduates.dropna(inplace=True)
Vallejo_graduates["UC_Acceptance_Rate"] = np.where(Vallejo_graduates["UC_GRADS"] > 0, (Vallejo_graduates["UC_GRADS"] / Vallejo_graduates["GRADS"]) * 100,0)
Vallejo_graduates["UC_Acceptance_Rate"] = Vallejo_graduates["UC_Acceptance_Rate"].round(2)
```

```python
display(Vallejo_graduates.head())
```

```python

# Reviewing first or Vallejo_graduates dataset

# Insight: Graduation and UC Graduation Totals by School and Ethnic Group

grad_data = Vallejo_graduates.groupby(["SCHOOL", "ETHNIC"]).sum()[["GRADS", "UC_GRADS"]].reset_index()

grad_data["GRADS to UC_GRADS Rate (%)"] = np.where(grad_data["UC_GRADS"] > 0, (grad_data["UC_GRADS"] / grad_data["GRADS"]) * 100,0)
grad_data["GRADS to UC_GRADS Rate (%)"] = grad_data["GRADS to UC_GRADS Rate (%)"].round(2)
grad_data.sort_values(by="GRADS to UC_GRADS Rate (%)", ascending=False, inplace=True)

grad_data.sample(5)

# uc_acceptance_by_school = Vallejo_graduates.groupby(['Year','SCHOOL'])['UC_Acceptance_Rate'].mean().reset_index()
# total_grads_by_ethnicity_gender = Vallejo_graduates.groupby(['ETHNIC', 'GENDER'])['GRADS'].sum().reset_index()

```

```python
UC_Acceptance_Rate_by_gender = Vallejo_graduates.groupby(['Year','ETHNIC','GENDER'])['UC_Acceptance_Rate'].median().reset_index()

UC_Acceptance_Rate_by_gender.sample(5)
```

```python
fig = px.scatter(
    UC_Acceptance_Rate_by_gender,
    x='Year',
    y='UC_Acceptance_Rate',
    color='ETHNIC',
    symbol='GENDER',
    size='UC_Acceptance_Rate',
    title="Median UC Acceptance Rate by Ethnicity and Gender over Time",
    labels={
        "UC_Acceptance_Rate": "UC Acceptance Rate (%)",
        "ETHNIC": "Ethnicity",
        "GENDER": "Gender",
        "Year": "Academic Year"
    },
    hover_name="ETHNIC",
    hover_data=["GENDER", "UC_Acceptance_Rate"]
)

# Update layout for readability
fig.update_layout(legend_title="Ethnicity", xaxis=dict(title="Year"), yaxis=dict(title="UC Acceptance Rate (%)"))

fig.show()
```

```python

ethnicity_color_map = {
    "African American": "#1f77b4",  # Blue
    "Asian": "#ff7f0e",  # Orange
    "Filipino": "#2ca02c",  # Green
    "Hispanic or Latino": "#d62728",  # Red
    "Pacific Islander": "#9467bd",  # Purple
    "White": "#8c564b",  # Brown
    "Two or More Races": "#e377c2",  # Pink
    "American Indian or Alaska Native": "#7f7f7f"  # Gray
}

school_color_map = {
    "Jesse M. Bethel High": ("#1f77b4", "#aec7e8"),  # Blue shades
    "John Finney High (Continuation)": ("#ff7f0e", "#ffbb78"),  # Orange shades
    "MIT Academy": ("#2ca02c", "#98df8a"),  # Green shades
    "Vallejo High": ("#d62728", "#ff9896")  # Red shades
}


fig = go.Figure()

for ethnicity in grad_data["ETHNIC"].unique():
    ethnicity_data = grad_data[grad_data["ETHNIC"] == ethnicity]
    color = ethnicity_color_map.get(ethnicity, "#1f77b4")

    fig.add_trace(go.Bar(x=ethnicity_data["SCHOOL"], y=ethnicity_data["GRADS"],
                         name=f"{ethnicity} - Total Graduates",
                         marker_color=color,
                         visible=True))

    fig.add_trace(go.Bar(x=ethnicity_data["SCHOOL"], y=ethnicity_data["UC_GRADS"],
                         name=f"{ethnicity} - UC-Eligible Graduates",
                         marker_color=color,
                         opacity=0.6,
                         visible=True))

for school in grad_data["SCHOOL"].unique():
    school_data = grad_data[grad_data["SCHOOL"] == school]
    grad_color, uc_grad_color = school_color_map.get(school, ("#1f77b4", "#aec7e8"))

    fig.add_trace(go.Bar(x=school_data["ETHNIC"], y=school_data["GRADS"],
                         name=f"{school} - Total Graduates",
                         marker=dict(color=grad_color),
                         visible=False))

    fig.add_trace(go.Bar(x=school_data["ETHNIC"], y=school_data["UC_GRADS"],
                         name=f"{school} - UC-Eligible Graduates",
                         marker=dict(color=uc_grad_color),
                         opacity=0.6,
                         visible=False))

fig.update_layout(
    updatemenus=[
        dict(
            buttons=[
                dict(label="View by School",
                     method="update",
                     args=[{"visible": [True] * (2 * len(grad_data["ETHNIC"].unique())) + [False] * (2 * len(grad_data["SCHOOL"].unique()))},
                           {"title": "Graduates and UC-Eligible Graduates by Ethnicity and School",
                            "xaxis_title": "School",
                            "yaxis_title": "Count"}]),

                dict(label="View by Ethnicity",
                     method="update",
                     args=[{"visible": [False] * (2 * len(grad_data["ETHNIC"].unique())) + [True] * (2 * len(grad_data["SCHOOL"].unique()))},
                           {"title": "Graduates and UC-Eligible Graduates by School and Ethnicity",
                            "xaxis_title": "Ethnicity",
                            "yaxis_title": "Count"}])
            ],
            direction="down",
            showactive=True,
            x=0.5,
            xanchor="center",
            y=1.15,
            yanchor="top"
        )
    ]
)

fig.update_layout(
    title="Graduates and UC-Eligible Graduates by School and Ethnicity via Select Box",
    xaxis_title="School",
    yaxis_title="Count",
    barmode="group",
    legend_title="Category",
    template="plotly_white"
)

fig.show()
```

```python
# Reviewing second or Vallejo_Enrollment_dropouts_by_school_district dataset

# Second Dataset
Solano_County_enrollment_dropouts_by_school_district = pd.read_csv("/content/drive/MyDrive/ClubStrideFolder/StudentData/Solano_County_enrollment_dropouts_by_school_district.csv")
Vallejo_Enrollment_dropouts_by_school_district = Solano_County_enrollment_dropouts_by_school_district[Solano_County_enrollment_dropouts_by_school_district['DISTRICT'].isin(['Vallejo City Unified'])]
Vallejo_Enrollment_dropouts_by_school_district['Latitude'] = Vallejo_Enrollment_dropouts_by_school_district['SCHOOL'].map(lambda school: coords.get(school, (None, None))[0])
Vallejo_Enrollment_dropouts_by_school_district['Longitude'] = Vallejo_Enrollment_dropouts_by_school_district['SCHOOL'].map(lambda school: coords.get(school, (None, None))[1])
Vallejo_Enrollment_dropouts_by_school_district.dropna(inplace=True)
display(Vallejo_Enrollment_dropouts_by_school_district.head())
```

```python


# Insight 1: Enrollment Statistics by Year
Dropouts_by_year = Vallejo_Enrollment_dropouts_by_school_district.groupby("Year")[["Dropouts_Grade_9", "Dropouts_Grade_10", "Dropouts_Grade_11", "Dropouts_Grade_12"]].sum()
Dropouts_by_year["Total High School Dropouts"] = Dropouts_by_year.sum(axis=1)

enrollment_by_year = Vallejo_Enrollment_dropouts_by_school_district.groupby("Year")[["Enrollment_Grade_9", "Enrollment_Grade_10", "Enrollment_Grade_11", "Enrollment_Grade_12"]].sum()
enrollment_by_year["Total High School Enrollment"] = enrollment_by_year.sum(axis=1)

# combined_data = enrollment_by_year.merge(Dropouts_by_year, on="Year")
# Calculate dropout rate
Dropouts_by_year["High School Dropout Rate (%)"] = (Dropouts_by_year["Total High School Dropouts"] / enrollment_by_year["Total High School Enrollment"]) * 100
Dropouts_by_year
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=Dropouts_by_year.index,
    y=enrollment_by_year["Total High School Enrollment"],
    mode='lines+markers',
    name='Total High School Enrollment',
    line=dict(color='royalblue', width=2),
    marker=dict(symbol="circle", color='royalblue', size=6),
    hovertemplate="Year: %{x}<br>Total Enrollment: %{y}",
))

fig.add_trace(go.Scatter(
    x=Dropouts_by_year.index,
    y=Dropouts_by_year["High School Dropout Rate (%)"],
    mode='lines+markers',
    name='Dropout Rate (%)',
    line=dict(color='seagreen', dash='dash', width=2),
    marker=dict(symbol="diamond", color='seagreen', size=6),
    yaxis="y2",
    hovertemplate="Year: %{x}<br>Dropout Rate: %{y}%",
))

fig.add_trace(go.Bar(
    x=Dropouts_by_year.index,
    y=Dropouts_by_year["Total High School Dropouts"],
    name='Total High School Dropouts',
    marker_color='crimson',
    # boxmean=True,
    # fillcolor='rgba(220,20,60,0.2)',
    hovertemplate="Year: %{x}<br>Total Dropouts: %{y}",
))

fig.update_layout(
    title=dict(
        text="Enrollment and Dropout Statistics by Year<br>",
        x=0.5,
        xanchor='center',
        font=dict(size=20)
    ),
    xaxis_title="Academic Year",
    yaxis=dict(
        title="Enrollment & Dropouts",
        titlefont=dict(color="royalblue"),
        tickfont=dict(color="royalblue")
    ),
    yaxis2=dict(
        title="Dropout Rate (%)",
        titlefont=dict(color="seagreen"),
        tickfont=dict(color="seagreen"),
        overlaying='y',
        side='right'
    ),
    template="plotly_white",
    legend=dict(
        title="Legend",
        orientation="h",
        x=0.5,
        xanchor="center",
        y=-0.25,
        bgcolor="rgba(255, 255, 255, 0.7)"
    ),
    hovermode="x unified",
    margin=dict(l=60, r=60, t=100, b=60)
)

fig.show()
```

```python
# Insight 2: Dropout Statistics by Year and Ethnicity
dropout_by_year_ethnicity = Vallejo_Enrollment_dropouts_by_school_district.groupby(["Year", "Ethnicity_Code"])["Total_Dropouts_Grades_9_12"].sum().unstack(fill_value=0)
enrollment_year_ethnicity = Vallejo_Enrollment_dropouts_by_school_district.groupby(["Year", "Ethnicity_Code"])["Total_Enrollment_Grades_9_12"].sum().unstack(fill_value=0)

# Calculate dropout rate
dropout_rate_by_year_ethnicity = (dropout_by_year_ethnicity / enrollment_year_ethnicity) * 100

dropout_rate_by_year_ethnicity
```

```python

fig = go.Figure()

for ethnicity in dropout_rate_by_year_ethnicity.columns:
    fig.add_trace(go.Bar(x=dropout_rate_by_year_ethnicity.index,
                             y=dropout_rate_by_year_ethnicity[ethnicity],
                            #  mode='lines+markers',
                             name=ethnicity,
                             hovertemplate="Year: %{x}<br>Dropout Rate: %{y:.2f}%"))

fig.update_layout(
    title="Dropout Rate by Ethnicity Over Time (Grades 9-12)",
    xaxis_title="Year",
    yaxis_title="Dropout Rate (%)",
    template="plotly_white",
    legend=dict(title="Ethnicity", orientation="h", x=0.5, xanchor="center", y=-0.2),
    hovermode="x unified"
)

fig.show()
```
